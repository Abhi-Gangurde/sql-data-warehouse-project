```python
#init 

catalog_name = "abhi_dwh_sql_based"
schema  = "raw"
source="/Workspace/Users/srinivas.adf90@gmail.com/abhi/sql-data-warehouse-project/datasets/"
sink = f"/Volumes/{catalog_name}/{schema}/volume"

azure_stg_account = "abhistgac1"
container_name = "devcontainer"
managed_location = f"abfss://{container_name}@{azure_stg_account}.dfs.core.windows.net/{catalog_name}"

#Create the catalog and schema and volumes 
spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog_name} MANAGED LOCATION '{managed_location}'")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema}")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog_name}.{schema}.volume")

dbutils.fs.cp(
    source,
    sink,
    recurse=True
)
```
