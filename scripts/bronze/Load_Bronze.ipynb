```python
%python
### Script Purpose :
"""
  Loads the raw data into th ronze table from exrernal csv files.
  Perform following actions:
    -Check if the bronze schema is present in the catalog , if not it will create it.
    -Truncate the brone tables before loading the data
    -Uses the spark api to read and write the data to the delta table

"""
```

```python
#init 

catalog_name = "abhi_dwh_sql_based"
schema  = "raw"
source = f"/Volumes/{catalog_name}/{schema}/volume/"
# source = f"/Workspace/Users/srinivas.adf90@gmail.com/abhi/sql-data-warehouse-project/datasets/"
```

```python
#prepare the prerequists for the bronze layer,

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog_name}.bronze")
print("Schema created --> ", catalog_name + ".bronze")
```

```python

INGESTION_CONFIG = [
    #Predefined configurations from source_crm location for the CRM data 
    {
        "source": "crm",
        "path": source + "source_crm/cust_info.csv",
        "table": "crm_cust_info"
    },
    {
        "source": "crm",
        "path": source + "source_crm/prd_info.csv",
        "table": "crm_prd_info"
    },
    {
        "source": "crm",
        "path": source + "source_crm/sales_details.csv",
        "table": "crm_sales_details"
    },

    #Predefined configurations from source_erp location for the ERP data
    {
        "source": "erp",
        "path": source + "source_erp/CUST_AZ12.csv",
        "table": "erp_cust_az12"
    },
    {
        "source": "erp",
        "path": source + "source_erp/LOC_A101.csv",
        "table": "erp_loc_a101"
    },
    {
        "source": "erp",
        "path": source + "source_erp/PX_CAT_G1V2.csv",
        "table": "erp_px_cat_g1v2"
    }
]
```

# Ingest Files Into Bronze Tables

```python
for item in INGESTION_CONFIG:
    print(f"Ingesting {item['source']}--> {catalog_name}.bronze.{item['table']}")

    #Read data from csv to the inmemory df 
    df = (
        spark.read.csv(item['path'], header=True, inferSchema=True)
    )

    #Write data into bronze as delta table
    df.write.mode("overwrite").saveAsTable(f"{catalog_name}.bronze.{item['table']}")
    print(f"Table {item['table']} created in {catalog_name}.bronze")
    print(f"=======================================================")
```
