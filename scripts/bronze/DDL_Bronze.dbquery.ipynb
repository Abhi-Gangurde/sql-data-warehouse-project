```sql
/*
=========================================================================
DDL Script : Create Bronze Tables 
=========================================================================
Script Purpose :
    This cript creates tables in the 'bronze' schema, dropping existing tables 
    if they already exists 
        Run this script to re-define the DDL structue of 'bronze' tables 
======================================================================

Why defining the DDL structure for every table is best practice ?
--> It ensures the data pipeline is reliable , governed and consistent.
    How?
        -->
            1. Schema Control (Avoid schema drift)
                If you dont define DDL, spark may infer schema automatically when reading files.
                    Problem: 
                        -->File Schema may change 
                        -->Column type may change 
                        -->New columns may appear unexpectedly 
            2. Data Quality enforcement
                DDL enforces correc data types and stucture.
                    Benifit:
                        -->Prevent invalid data
                        -->Ensures correct transformation downstream
            3. Faster Query Performance:
                Explicit schema allows spark to skip schema inference 
                    Without DDL:
                        -->Spark scans files to infer schema 
                        -->Extra overhead during job execution
                    With DDL:
                        -->Schema already known 
                        -->Faster job startup
            4. Data Governance & catalog Management 
                With system like one catalog or unity catalog, defining tables using DDL enables:
                    -->Data lineage 
                    -->Access control
                    -->Table ownership
                    -->Column level security 
            5. Standardize Across Layers 
                EA layer has clear contract
                    -->Bronze : Raw schema (minimal transformation)
                    -->Silver : cleaned and standardized (correct dtypes , validated fields )
                    --> Gold  : Business Ready analytics table (Agregations , KPIss , Dimensional models)
            6. Prevent Pipeline failures
                Explicit schemas prevent runtime surprises 
            7. Better version control 
                -->Track Schema changes 
                --> Rollback if needed 
                --> CI/CD compatibility 
            8. Support for delta features 
                DDL enables advance features of delta lake
                    -->Schema enforcement 
                    -->Schema evolution
                    -->Time travel 
                    -->Contraints 

======================================================================

*/

USE CATALOG abhi_dwh_sql_based;


-- Create schema if not 
CREATE SCHEMA IF NOT EXISTS bronze;

USE SCHEMA bronze;


-- Ensure pipeline starts with clean table 

DROP TABLE IF EXISTS bronze.crm_cust_info;

DROP TABLE IF EXISTS bronze.crm_prd_info;

DROP TABLE IF EXISTS bronze.crm_sales_details;

DROP TABLE IF EXISTS bronze.erp_loc_a101;

DROP TABLE IF EXISTS bronze.erp_cust_az12;

DROP TABLE IF EXISTS bronze.erp_px_cat_g1v2;

--Create tables schema

CREATE TABLE bronze.crm_cust_info(
    cst_id          varchar(50),
    cst_key         varchar(50),
    cst_firstname   VARCHAR(50),
    cst_lastname    VARCHAR(50),
    cst_marital_status VARCHAR(50),
    cst_gndr        VARCHAR(50),
    cst_create_date Date    
    );


CREATE TABLE bronze.crm_prd_info (
    prd_id       INT,
    prd_key      VARCHAR(50),
    prd_nm       VARCHAR(50),
    prd_cost     INT,
    prd_line     VARCHAR(50),
    prd_start_dt DATE,
    prd_end_dt   DATE
);


CREATE TABLE bronze.crm_sales_details (
    sls_ord_num  VARCHAR(50),
    sls_prd_key  VARCHAR(50),
    sls_cust_id  INT,
    sls_order_dt INT,
    sls_ship_dt  INT,
    sls_due_dt   INT,
    sls_sales    INT,
    sls_quantity INT,
    sls_price    INT
);



CREATE TABLE bronze.erp_loc_a101 (
    cid    VARCHAR(50),
    cntry  VARCHAR(50)
);


CREATE TABLE bronze.erp_cust_az12 (
    cid    VARCHAR(50),
    bdate  DATE,
    gen    VARCHAR(50)
);


CREATE TABLE bronze.erp_px_cat_g1v2 (
    id           VARCHAR(50),
    cat          VARCHAR(50),
    subcat       VARCHAR(50),
    maintenance  VARCHAR(50)
);
```
